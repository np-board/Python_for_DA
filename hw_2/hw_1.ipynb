```python
def check_float(arg) -> bool:
  try:
    if arg > 0.0:
      return True
    else:
      raise ValueError('Ошибка! Цена меньше или равна нулю!')
  except TypeError:
    print(
        f'Ошибка! Цена не является типом данных float, будет выставлена цена 1 рубль! Тип:{str(type(arg))}')
    return False
  except ValueError:
    print('Ошибка! Цена меньше или равна нулю!')
    return False

def check_str(arg:str) -> bool:
  try:
    if type(arg) is str:
      return True
    else:
      raise TypeError()
  except TypeError:
    print('Ошибка! В качестве добавки введена не строка')
    return False

class Chocolate:

  def __init__(self, price:float, additive: str = ''):

    if check_float(price) is True:
        self.price = price
    else: self.price = 1

    if check_str(additive) is True:
      self.additive = additive
    else: self.additive = ''

  def __str__(self):

    return f'Шоколадный батончик, {str(self.price)} рублей'

  def show_my_chocolate(self):
      if self.additive.strip():
        return f'Шоколадный батончик, добавка: {self.additive}'
      else:
        return 'Обычный шоколадный батончик'

shelf = {
    1 : Chocolate(1.2, ' '),
    2 : Chocolate(1.2, 'Солёный огурчик'),
    3 : Chocolate(2,-3),
    4 : Chocolate('Мясной мёд',2)
}

for index, value in shelf.items():
  print(f'{str(index)} : {value} — {value.show_my_chocolate()}')
```
